### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
# 1) numeri elevati al quadrato

# ciclo
numeri  = [1,2,3,4,5]
quadrati = []
for n in numeri:
    quadrati.append(n**2)
print(f"quadrati ciclo: {quadrati}")

# list comprehension
quadrati_comp = [n**2 for n in numeri]
print(f"quadrati list comp.: {quadrati_comp}")

# 2) numeri pari tra 0 e k

# ciclo
k = 20
pari = []
for n in range(k):
    if n % 2 == 0:
        pari.append(n)
print(f"pari ciclo: {pari}")

# list comprehension
pari_comp = [n for n in range (k) if n % 2 == 0]
print(f"pari list comp.: {pari_comp}")

# 3) lunghezze delle parole

# ciclo
parole = ["python", "ciao", "programma", "ok"]
lunghezze = []
for parola in parole:
    lunghezze.append(len(parola))
print(f"lunghezze ciclo: {lunghezze}")

# list comprehension
lunghezze_comp = [len(parola) for parola in parole]
print(f"lunghezze list comp.: {lunghezze_comp}")
```

### **2. Vettore di Numeri Primi**

1. **Crea un vettore** contenente tutti i numeri primi compresi tra **0 e 10**  
   _(Puoi scriverli direttamente nell'array)._

2. **Conta quanti numeri ci sono nel vettore** utilizzando la funzione `len()`.  **Ottieni lo stesso numero** accedendo all'attributo `.size` del vettore.

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
import numpy as np

# funzione che controlla se un numero è primo
def is_prime(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    i = 3
    while i * i <= n:   
        if n % i == 0:
            return False
        i += 2          # salta i pari
    return True

# 1.
a = np.array([1,2,3,5,7])

# 2.
print(f"lunghezza con len: {len(a)}")
print(f"lunghezza con size: {a.size}")

# 3.
print(f"tipo del vettore: {a.dtype}")

# 4.
primes = np.array([n for n in range(2, 10) if is_prime(n)])

print(f"numeri primi: {primes}")
```

### **3.  Operazioni su array**
Crea un array **a**, 1D (senza digitarlo esplicitamente), e genera due nuovi array **b**, contenente una sottostringa a piacere e **c** che è il reverse dell'array **a** (esempio [1,2,3] diventa [3,2,1]) .  Dividi l'array **a** per l'array **c**. Fai la stessa cosa per una lista.

```python
# versione con gli array
a = np.arange(1, 20)
print(f"array a: {a}")

b = a[5:15]
print(f"sottoarray di a: {b}")

c = a[::-1]
print(f"reverse di a: {c}")

d = a/c
print(f"a/c: {d}")

# versione con le liste
listA = list(range(1,20))
print(f"lista A: {listA}")

listB = listA[6:12]
print(f"sottolista di A: {listB}")

listC = listA[::-1]
print(f"reverse di A: {listC}")

listD = listA/listC
print(f"A/C: {listD}")
# non si possono dividere due liste
```
